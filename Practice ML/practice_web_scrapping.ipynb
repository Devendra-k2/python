```python
import requests
from bs4 import BeautifulSoup

web_data= requests.get('https://www.scrapethissite.com/pages/',verify=False)
# print(web_data)
```

```python
# print(web_data.text)
soup=BeautifulSoup(web_data.text)
print(soup.prettify())
```

```python
job_cards=soup.find('div',class_='row')
# job_cards=soup.select('div.md\\:col-span-6')
print(job_cards)
```

```python
job_cards2=soup.find_all('h1')
print(job_cards2)
```

```python
# job_cards3=soup.find_all('a')
title=job_cards.find('h3',class_="page-title").text.replace(' ','')

print(title)
```

```python
content=job_cards.find('p',class_='lead session-desc').text.replace(' ','')
print(content)
```
