```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url_data=requests.get('https://www.scrapethissite.com/pages/',verify=False).text

# getting data in source code form 

source_data=BeautifulSoup(url_data)

whole_data=source_data.find_all('div',class_='page')
# print(whole_data)

title_list=[]
content_list=[]

for data in whole_data:
    title=data.find('h3').text.lstrip()
    content=data.find('p',class_='lead').text.lstrip()
    title_list.append(title)
    content_list.append(content)

print(len(title_list))
print(len(content_list))
```

```python
rows = []

for data in whole_data:
    title = data.find('h3').get_text(strip=True)
    content = data.find('p', class_='lead').get_text(strip=True)

    rows.append({
        "Title": title,
        "Content": content
    })

# import pandas as pd
df = pd.DataFrame(rows)
print(df)
```
